## 1. Extract unique values

```python
d = {'gfg' : [5,6,7,8],' is' : [10,11,7,5], 'best' : [6,12,10,8], 'for' : [1,2,5]}

L = []
for i in d.values():
    L.append(i)
print(list(set([x for row in L for x in row])))
```

```python
import numpy as np
d = {'gfg' : [5,6,7,8],' is' : [10,11,7,5], 'best' : [6,12,10,8], 'for' : [1,2,5]}
L = []
for i in d.values():
    L.append(i)
print(L)
flat = np.array(L).flatten().tolist() # cannot be used for irregular list(non-homogenous list)
print(flat)
```

## 4. Sort list of dict by values

```python
from operator import itemgetter
d = [
    {"name": "Nandini", "age": 20},
    {"name": "Manjeet", "age": 20},
    {"name": "Nikhil", "age": 19}
]
print("Sorted : ",sorted(d, key= itemgetter('age','name'), reverse=False))
```

```python
# Using lambda fxn
d = [
    {"name": "Nandini", "age": 30},
    {"name": "Manjeet", "age": 20},
    {"name": "Nikhil", "age": 19}
]
print(sorted(d,key= lambda i: i['age']))
```

## 7. Insertion at the beginning in OrderedDict

```python
from collections import OrderedDict
d = OrderedDict([('a',1), ('b',2), ('c',3)])
d.update({'d':4})
d.move_to_end('d',last = False)
print(d)
```

## 10. Anagrams

```python
a = ['bat', 'nat', 'tan', 'ate', 'eat', 'tea']
res = {}

for word in a:
    key = "".join(sorted(word))
    res[key] = res.get(key,[]) + [word]
print(list(res.values()))
```

```python
from collections import defaultdict

a = ['bat', 'nat', 'tan', 'ate', 'eat', 'tea']
res = defaultdict(list)
for word in a:
    res["".join(sorted(word))].append(word)

print(list(res.values()))
```
