## 1. Get size of set

```python
import sys
set1 = {1,"A",2,"B"}
print("Size of set",sys.getsizeof(set1),"bytes")
```

## 2. Iterating over set

```python
a = set("Vedant")

for i in a:
    print(i)
```

```python
a = set("Mercedes")

for i in a.__iter__():
    print(i)
```

## 3. Max and Min

```python
s = {3,45,64,32,54,76,23}
print("Max :",max(s))
print("Min :",min(s))
```

## 4. Removing item

```python
a = {2,3,4,5,6,7,1}

a.discard(1)
a.remove(5)
print(a)
```

## 5. Set into tuple and tuple into set

```python
a = set("hello")
b = set("Germany")

print("Set :",a)
print("Set :",b)
t = tuple(a)
print("Tuple :",t)

t1 = (1,2,3,4,5)
print("Tuple :",t1)
s = set(t1)
print("Set :",s)
```
