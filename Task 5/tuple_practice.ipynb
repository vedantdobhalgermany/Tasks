## 2. Max and Min K elements in tuple

```python
# Method 1 - slicing + sorted
t = (2,4,3,6,8,11,1,9)
K = 2

temp = sorted(t)
mi = temp[:K]
ma = temp[-K:]

print(mi)
print(ma)
```

```python
# Method 2 - sorted + loop
t = (2,4,3,6,8,11,1,9)
K = 2

l = sorted(t)
mi,ma = [],[]

for i,val in enumerate(l):
    if i<K:
        mi.append(val)
    if i>= len(l)-K:
        ma.append(val)

print(mi)
print(ma)
```

## 3. Create a list of tuple with numbers and their cubes

```python
# Method 1 - Using list comprehension
a = [1,2,3,4,5]
res = [(n,n**3) for n in a]
print(res)
```

```python
# Method 2 - Using map with lambda
a = [1,2,3,4,5]
res = list(map(lambda n:(n,n**3),a))
print(res)
```

```python
# Method 3 - Using zip
a = [1,2,3,4,5]
res = list(zip(a,[n**3 for n in a]))
print(res)
```

## 4. Adding tuple to list and vice-versa

```python
a = [1,2,3]
b = (4,5)
c = [6,7]

# add tuple to list
a.extend(b)
print(a)

# add list to tuple b
d = b+ tuple(c)
print(d)
```

## 5. Closest pair to Kth index element in tuple

```python
# Method 1 - Using min with lambda
t = [(3,4,9),(5,6,7)]
tup = (1,2,5)
K = 3

res = min(t, key=lambda x: abs(x[K-1] - tup[K-1]))
print(res)
```

```python
# Method 2 - using sorted() with custom key
t = [(3,4,9),(5,6,7)]
tup = (1,2,5)
K = 3

s = sorted(t, key = lambda x: abs(x[K-1] - tup[K-1]))
print(s[0])
```

## 6. Join tuples if similar initial elements

```python
t = [(5,6),(5,7),(6,10),(6,8)]
temp = {}

for x in t:
    temp[x[0]] = temp.get(x[0],[]) + list(x[1:])

res = [(k,) + tuple(v) for k,v in temp.items()]
print(res)
```

## 7. Extract digits from tuple list

```python
t = [(15,3),(3,9),(1,10),(99,2)]
temp = ''.join([str(i) for s1 in t for i in s1])
res = [int(i) for i in set(temp)]
print(res)
```

## 8. All pair combinations of 2 tuples

```python
t1 = (1,2)
t2 = (3,4)

res = [(a,b) for a in t1 for  b in t2] + [(a,b) for a in t2 for b in t1]
print(res)
```

## 9. Remove tuples of length K

```python
t = [(1,2),(3,4),(4,5,6),(7,),(8,9,0)]
K = 2
res = [tup for tup in t if len(tup) != K]
print(res)
```

## 10. Sort a list of tuples by second item

## 11. Order tuple by list

```python
t = [('Gfg', 3), ('best', 9), ('CS', 10), ('Geeks', 2)]
l = ['Geeks', 'best', 'CS', 'Gfg']

temp = dict(t)
res = [(key, temp[key]) for key in l]
print(res)
```

## 12. Flatten tuple of list to tuple

```python
from itertools import chain
tup = ([5,6],[6,7,8,9],[3])
res = tuple(chain.from_iterable(tup))
print(res)
```

## 13. Convert Nested Tuple to Custom Key Dictionary

```python
a = ((4, 'Gfg', 10), (3, 'is', 8), (6, 'Best', 10))
keys = ['key', 'value', 'id']
res = [dict(zip(keys, v)) for v in a]
print(res)
```
