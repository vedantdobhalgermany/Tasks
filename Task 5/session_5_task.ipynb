###`Q1:` Join Tuples if similar initial element
While working with Python tuples, we can have a problem in which we need to perform concatenation of records from the similarity of initial element. This problem can have applications in data domains such as Data Science.

For eg.
```
Input  : test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]
Output : [(5, 6, 7, 8), (6, 10), (7, 13)]
```

````python
from setuptools._distutils.compilers.C import unix

test_list = [(5,6),(5,7),(5,8),(6,10),(7,13)]

temp = []

for x in test_list:
    temp.append(x[0])
unique = set(temp)

result = []
for i in unique:
    result.append([i])
    for j in test_list:
        if j[0] == i:
            result[-1].append(j[1])

print(list(map(tuple,result)))
````

###`Q2:` Multiply Adjacent elements (both side) and take sum of right and lest side multiplication result.


For eg.
```
The original tuple : (1, 5, 7, 8, 10)
Resultant tuple after multiplication :

(1*5, 1*5+5*7, 7*5 + 7*8, 8*7 + 8*10, 10*8) -> (5, 40, 91, 136, 80)

output-(5, 40, 91, 136, 80)
```

````python
t = (1,5,7,8,10)

res = []

res.append(t[0]*t[1])
for i in range(1,len(t)-1):
    res.append(t[i]*t[i+1] + t[i]*t[i-1])
res.append(t[-1]*t[-2])

print(tuple(res))
````

###`Q3`: Check is tuples are same or not?
Two tuples would be same if both tuples have same element at same index
```
t1 = (1,2,3,0)
t2 = (0,1,2,3)

t1 and t2 are not same
```

````python
t1 = (1,2,3,0)
t2 = (1,2,3,9)

print(len(t1) == len(t2) and all(i == j for i,j in zip(t1, t2)))
````

````python
t1 = (1,2,3,0)
t2 = (1,2,3,9)

flag = 1
if len(t1) == len(t2):
    for i in range(len(t1)):
        if t1[i] != t2[i]:
            flag = 0
            break
else:
    flag = 0


if flag == 1:
    print("True")
elif flag == 0:
    print("False")
````

###`Q4`: Count no of tuples, list and set from a list
```
list1 = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi', 'bye'],['a', 'b']]

```
`Output:`

```
List-2
Set-2
Tuples-1
```

````python
list1 = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi', 'bye'],['a', 'b']]

l,t,s,d= 0,0,0,0
for i in list1:
    if type(i) == list:
        l += 1
    elif type(i) == tuple:
        t += 1
    elif type(i) == set:
        s += 1
    elif type(i) == dict:
        d += 1
print("List :",l)
print("Tuple :",t)
print("Set :",s)
print("Dictionary :",d)
````

###`Q5`: Shortlist Students for a Job role
Ask user to input students record and store in tuples for each record. Then Ask user to input three things he wants in the candidate- Primary Skill, Higher Education, Year of Graduation.

Show every students record in form of tuples if matches all required criteria.

It is assumed that there will be only one primry skill.

If no such candidate found, print `No such candidate`

`Input:`
```
Enter No of records- 2
Enter Details of student-1
Enter Student name- Manohar
Enter Higher Education- B.Tech
Enter Primary Skill- Python
Enter Year of Graduation- 2022
Enter Details of student-2
Enter Student name- Ponian
Enter Higher Education- B.Sc.
Enter Primary Skill- C++
Enter Year of Graduation- 2020

Enter Job Role Requirement
Enter Skill- Python
Enter Higher Education- B.Tech
Enter Year of Graduation- 2022
```

`Output`
```
('Manohar', 'B.tech', 'Python', '2022')
```


````python
students = []

num = int(input("Enter the number of candidates :"))
for i in range(num):
    print("Enter details of ",i+1, " candidate :")
    name = input("Enter candidate name:")
    h_ed = input("Enter the higer education:")
    p_skill = input("Enter the primary skill:")
    yog = input("Enter the year_of_graduation:")

    students.append((name,h_ed,p_skill,yog))

req_skill = input("Enter the required skill:")
req_hed = input("Enter required higer education:")
req_yog = input("Enter required year of graduation:")

flag = False
for i in students:
    if i[1]== req_hed and i[2]== req_skill and i[3]== req_yog:
        print(i)
        flag = True
if flag == False:
    print("No such candidate")
````

```
Input : ar1 = [1, 5, 10, 20, 40, 80]
        ar2 = [6, 7, 20, 80, 100]
        ar3 = [3, 4, 15, 20, 30, 70, 80, 120]

Output : [80, 20]
```
