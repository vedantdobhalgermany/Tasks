## 1. Palindrome check

```python
s = "malayalam"

if  s == s[::-1]:
    print("Yes")
else:
    print("No")
```

```python
s = "geeks"
rev = ''.join(reversed(s))

if s == rev:
    print("Yes")
else:
    print("No")
```

## 2. Check Whether the String is Symmetrical or Palindrome

```python
s = "abaaba"

half = len(s)//2
sym = s[:half] == s[half:] if len(s)%2==0 else s[:half]==s[half+1:]
pal = s==s[::-1]

print("Symmetrical" if sym else "Non Symmetrical")
print("Palindrome" if pal else "Not Palindrome")
```

## 3. Reverse Words in a Given String

```python
s = "Python is fun"
res = ' '.join(s.split()[::-1])
print(res)
```

## 5. Check if String Contains Substring

```python
s = "Geeks for geeks"

if "for" in s:
    print("Yes")
else:
    print("No")
```

```python
# using find -> most suitable method

s = "Geek for geeks"
s2 = "for"

if s.find(s2) != -1:
    print("Yes")
else:
    print("No")
```

```python
# using operator.contains()
import operator as op

s = "Geeks for geeks"
s2 = "Geeks"

if op.contains(s,s2):
    print("Yes")
else:
    print("No")
```

## 9. Print even length words in a string

```python
s = "Python language is very beautiful"
words = s.split()
even_words = [w for w in words if len(w)%2==0]
res = " ".join(even_words)
print(res)
```

## 10. Accept the Strings Which Contains all Vowels

```python
s = "geeksforgeeks"
v = set('aeiou')
if v.issubset(set(s.lower())):
    print("Yes")
else:
    print("No")
```

```python
# using all
s = "vedantDobhal"
v = "aeiou"
if all(i in s.lower() for i in v):
    print("Yes")
else:
    print("No")
```

## 11. Count the Number of matching characters in a pair of string

```python
# Set intersection
s1 = "apple"
s2 = "grape"

res = len(set(s1.lower()).intersection(set(s2.lower())))
print(res)
```

```python
s1 = "apple"
s2 = "grape"
res = len([c for c in set(s1.lower()) if c in set(s2.lower())])
print(res)
```
