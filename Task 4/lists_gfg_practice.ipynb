## 4. Reversal Algorithm for Array Rotation

```python
arr = [1, 2, 3, 4, 5, 6, 7]
d = 2
n = len(arr)

# Reverse first d elements
arr[:d] = reversed(arr[:d])

# Reverse remaining elements
arr[d:] = reversed(arr[d:])

arr.reverse()
print(arr)
```

## 6. Find Remainder of Array Multiplication Divided by N

```python
arr = [100, 10, 5, 25, 35, 14]
n = 11
mul = 1
for i in arr:
    mul = i*mul
mul = mul % n
print(mul)
```

## 7. Check if given array is Monotonic

```python
arr = [6, 5, 4, 4]
n = len(arr)
inc = all(arr[i] <= arr[i+1] for i in range(n-1))
dec = all(arr[i] >= arr[i+1] for i in range(n-1))
print(inc or dec)
```

```python
# using zip
arr = [6, 5, 4, 4]
inc = all(a<=b for a, b in zip(arr, arr[1:]))
dec = all(a>=b for a, b in zip(arr, arr[1:]))
print(inc or dec)
```

```python
arr = [6, 5, 4, 4]
inc = sorted(arr) == arr
dec = sorted(arr,reverse=True) == arr
print(inc or dec)
```

## 11. Different ways to Clear a List in Python

```python
# using clear method
a = [1, 2, 3, 4, 5]
a.clear()
print(a)
```

```python
# using del
a = [1,2,3,4,5]
del a[:]
print(a)
```

```python
# using pop
a = [1,2,3,4,5]
while a :
    a.pop()
print(a)
```

```python
# reassigning to []
a = [1,2,3,4,5]
a = []
print(a)
```

## 18. Remove Multiple Elements from List
